```python
class VacuumCleaner:
    def __init__(self):
        self.location = "A"
        self.rooms = {
            "A": "Dirty",
            "B": "Dirty"
        }

    def simple_reflex_agent(self):
        status = self.rooms[self.location]

        if status == "Dirty":
            print("Action: Pick the Dust")
            self.rooms[self.location] = "Clean"

        elif self.location == "A":
            print("Action: Move Right")
            self.location = "B"

        elif self.location == "B":
            print("Action: Move Left")
            self.location = "A"

    def goal_based_agent(self):
        if self.rooms["A"] == "Clean" and self.rooms["B"] == "Clean":
            print("Goal Achieved: Both rooms are clean!")
            return True

        if self.rooms[self.location] == "Dirty":
            print("Action: Pick the Dust")
            self.rooms[self.location] = "Clean"

        elif self.location == "A":
            print("Action: Move Right")
            self.location = "B"

        elif self.location == "B":
            print("Action: Move Left")
            self.location = "A"

        return False


vacuum = VacuumCleaner()

print("Initial State:")
print(vacuum.rooms)

print("\n--- Simple Reflex Agent ---")

for i in range(4):
    print("\nCurrent Location:", vacuum.location)
    vacuum.simple_reflex_agent()
    print("Room Status:", vacuum.rooms)

vacuum = VacuumCleaner()

print("\n--- Goal Based Agent ---")

while True:
    print("\nCurrent Location:", vacuum.location)
    print("Room Status:", vacuum.rooms)

    if vacuum.goal_based_agent():
        break
```
